```python
import os
import xml.etree.ElementTree as ET
import numpy as np # linear algebra
import pandas as pd # data processing, CSV file I/O (e.g. pd.read_csv)
import torch 
from torchvision import transforms 
from PIL import Image
import random



```

```python
# localize the components 
def prepare_data(base_path):
    all_boards = []
    
    
    for board_folder_name in os.listdir(base_path):
        folder_path = os.path.join(base_path, board_folder_name)
        
        
        if not os.path.isdir(folder_path):
            continue
            
        board_data = {'name': board_folder_name, 'components': {}}
        
     
        img_path = os.path.join(folder_path, f"{board_folder_name}.jpg")
        xml_path = os.path.join(folder_path, f"{board_folder_name}.xml")
        
       
        if os.path.exists(img_path) and os.path.exists(xml_path):
            img = Image.open(img_path).convert('RGB')
            tree = ET.parse(xml_path)
            root = tree.getroot()
            
            
            for obj in root.findall('object'):
                
                full_name = obj.find('name').text
                label = full_name.split()[0].lower() 
                
               
                bbox = obj.find('bndbox')
                coords = (
                    int(bbox.find('xmin').text),
                    int(bbox.find('ymin').text),
                    int(bbox.find('xmax').text),
                    int(bbox.find('ymax').text)
                )
                
               
                crop = img.crop(coords)
                
                # Organising them on categories 
                if label not in board_data['components']:
                    board_data['components'][label] = []
                board_data['components'][label].append(crop)
                
            all_boards.append(board_data)
            
    return all_boards

    
                
                
            
```
